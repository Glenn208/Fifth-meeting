Scipy

````python
from scipy import stats
````

````python
distribusi_normal = stats.norm.rvs(
    loc= 70, #mean
    scale=5, #std
    size=10000 #banyakdata
)

plt.hist(distribusi_normal, bins=30)
plt.title("Distribusi Normal")
plt.xlabel("Nilai")
plt.ylabel("frekuensi")
plt.show
````

````python
distribusi_exponential = stats.expon.rvs(
    scale=5, #std
    size=10000 #banyakdata
)

plt.hist(distribusi_exponential, bins=30)
plt.title("Distribusi Exponential")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show
````

````python
distribusi_poisson = stats.poisson.rvs(
    mu=5, #eata-rata
    size=10000 #banyakdata
)

plt.hist(distribusi_poisson, bins=30)
plt.title("Distribusi poisson")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show
````
