```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import weibull_min
# x values for the theoretical PDF
x = np.linspace(0, 3000, 500)

# Parameter Weibull
k = 2
lambda_ = 1000
# Membuat 100 data Weibull
data = lambda_ * np.random.weibull(k, 100)
print(data)
```

```python
# Calculate Weibull PDF
pdf = weibull_min.pdf(x, k, scale=lambda_)

# Plot histogram
plt.hist(data, bins=10, density=True,
         alpha=0.5, edgecolor='blue',
         label='Histogram')

# Plot theoretical PDF
plt.plot(x, pdf, 'r-', linewidth=2,
         label='PDF')

plt.xlabel('Lifetime')
plt.ylabel('Density')
plt.title('Histogram & PDF')
plt.legend()
plt.show()
```

```python
x = np.sort(data)
y = np.arange(1, len(x)+1) / len(x)

plt.plot(x, y)
plt.xlabel("Lifetime")
plt.ylabel("Cumulative Probability")
plt.title("CDF")
plt.grid()
plt.show()
```
